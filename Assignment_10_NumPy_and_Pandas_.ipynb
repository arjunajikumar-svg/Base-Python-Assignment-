#Numpy Array Operations

```python
import numpy as np
```

## 1. Create a 1D NumPy Array

```python
temperatures_w1 = np.array ([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:",temperatures_w1)
```

##2. Inspection and Properties

```python
print("Shape of the Array :",temperatures_w1.shape)
print("Data type of Array :",temperatures_w1.dtype)
print("Number of Elements :",temperatures_w1.size)
```

##3. Array Operations


```python
temperatures_f = (temperatures_w1* 9/5) + 32

print("Temperatures in Fahrenheit :",temperatures_f)
```

```python
print("Maximum Temperature :",temperatures_w1.max())
print("Minimum Teprarature :",temperatures_w1.min())
print("Mean Teprarature :",temperatures_w1.mean())
```

##4. Array Slicing and Indexing

```python
print("Temperatures for the 3 days :",temperatures_w1[:3])

print("Temperatures for the weekend :",temperatures_w1[-2:])

print("Temperatures for the middle 3 days :",temperatures_w1[2:5])
```

## 5. Create a 2D Array

```python
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])

print("2D Temperature Array :\n",temperatures)
print("\n","Dimension of the Array :",temperatures.ndim)
```

```python
print("Shape of the Array :",temperatures.shape)
print("Data type of Array :",temperatures.dtype)
print("Number of Elements :",temperatures.size)
```

```python
print("Week 1 :",temperatures[0])
print("Week 2 :",temperatures[1])

print("\n","The temperatures for each week and the weekends :\n",temperatures[:,-2:])
```

#Pandas Series

```python
import pandas as pd
```

##1. Creating Pandas Series

```python
marks = pd.Series([ 95, 92, 89, 85, 80],index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

##2. Indexing and Slicing:

```python
print("1st rank student : ", marks.iloc[0])
```

```python
print("Top 3 Rank \n")
print(marks.loc[["Rank1","Rank2","Rank3"]])
```

```python
print("3rd rank student mark : ", marks.iloc[2])
```

```python
print(marks[marks>90])
```

##3. Manipulating Series

```python
marks["Rank1"]=100
print(marks)
```

```python
marks = marks.drop("Rank5")
print(marks)
```

```python
cgpa = marks/10
print(cgpa)
```

#Pandas DataFrame

##1. Creating Pandas DataFrame

```python
transactions = pd.DataFrame ({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
     'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

##2. Data Exploration

```python
print("Head:\n",transactions.head())
print("Tail:\n",transactions.tail())
print('Shape:\n',transactions.shape)
print('Column names:\n',transactions.columns)
print('Data types:\n',transactions.dtypes)
```

```python
print(transactions[["ProductCategory","Amount"]])
```

```python
print(transactions.iloc[:,-3:])
```

```python
filter_data = transactions[(transactions['Region']=='North')&(transactions['Amount'] >200)]
print(filter_data)
```

```python
count = transactions["ProductCategory"].value_counts()
print(count)
```

```python
print(transactions["Region"].unique())
```

```python
mean_amount = transactions.groupby("Region")["Amount"].mean()
print(mean_amount)
```

##3. Manipulating the DataFrame


```python
transactions.loc[transactions["TransactionID"] ==102,"Amount"] = 165
print(transactions)
```

```python
transactions["Discount"] = transactions["Amount"]/10

print(transactions)
```

```python
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)
```

```python
transactions = transactions.drop(columns="Discount")

print(transactions)
```
