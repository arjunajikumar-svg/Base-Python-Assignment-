#While loop & Control Statements :

*1. Set Up the Game*

```python
import random
```

```python
secret_number = random.randint(1, 10)
print(secret_number)
```

*2. Prompt the User*

```python
# Ask the user to guess the number.
attempts = 3

print("Welcome to the Number Guessing Game")
print("I have a number between 1 and 10.")
print(f"You have {attempts} chances to guess it.")
```

*3. Implement the Guessing Logic*

*4. Control Statements*

```python


while attempts > 0:
    guess = int(input(" Guess the number: "))

    if guess < 1 or guess > 10:
        print("  Out of range Guess between 1 and 10.")
        continue

    attempts -= 1

    if guess == secret_number:
        print(f" Correct The number was {secret_number}")
        break
    elif guess < secret_number:
        print(f" Too low Attempts left: {attempts}")
    else:
        print(f" Too high Attempts left: {attempts}")
else:
    print(f" Better luck next time Answer: {secret_number}")

```

#Function
*BMI Calculator*

```python
def calculate_bmi(weight, height):
  return round(weight/(height**2),2)

weight = float(input('Enter your weight in (Kg): '))
height = float (input(' Enter your height in (Cm): '))
bmi = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi}")
```
